# Collect masks in different folders and renumber them
After applying sam on different partitions of the movie, place the results in folders in alphabetical order. This script collect the results and put them in one folder

```python
import os
import shutil
```

```python
# Define the parent directory containing all result folders
parent_dir =  '/mnt/DATA/Mahsa/movies/LongRecordings/2025_03_19/Confocal_devided/Masks/'


# Define the target directory where all images will be consolidated
target_dir =  '/mnt/DATA/Mahsa/movies/LongRecordings/2025_03_19/Confocal_devided/Masks_all/'
```

```python
# Create the target directory if it doesn't exist
os.makedirs(target_dir, exist_ok=True)

# Initialize a counter for naming images sequentially
image_counter = 0
entries = os.listdir(parent_dir)
folders = [entry for entry in entries if os.path.isdir(os.path.join(parent_dir, entry))]
folders.sort()
print(folders)
# Iterate through each folder in the parent directory
for folder_name in folders:
    folder_path = os.path.join(parent_dir, folder_name)    
    folder_path = os.path.join(folder_path, folder_name + '_masks') 
    # Ensure we're working with a directory
    if os.path.isdir(folder_path):
        print(folder_name)
        image_files = [file_name for file_name in os.listdir(folder_path)
                       if file_name.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff'))]
        image_files.sort()
        for file_name in image_files:
            source_file = os.path.join(folder_path, file_name)
            # Define the new file name with the appropriate extension
            new_file_name = f'frame_{str(image_counter).zfill(5)}{os.path.splitext(file_name)[1]}'
            destination_file = os.path.join(target_dir, new_file_name)
            # Move and rename the file
            shutil.copy(source_file, destination_file)
            image_counter += 1

print(f'All images have been consolidated and renamed in {target_dir}.')
```
