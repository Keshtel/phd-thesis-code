# 1. Steps to correct the  csv file after skletonization if some frames are missing
Use the file generated by `B1_ConvertSAM2DLC/ConvertToSkleton/skletonize.ipynb` script called `results.csv`


```python
csvFile_address = '/mnt/DATA/Mahsa/movies/LongRecordings/2025_03_19/sklet2DLC/results.csv'
csvFile_Out = csvFile_address.split(".")
csvFile_Out = csvFile_Out[0]
rows,rowsArr_float = extract_coordinates(csvFile_address,numx=0,numy=0)#numx=3,numy=1)
```

```python
#find the frames where the skleton is missing
length = 11900# real number of frames
existence = np.ones((length))
for t in range(len(rows)):
    string = rows[t][0]
    ind = int(string)
    existence[ind] = 0
```

```python
#add zero for skleton of missing frames
for i in np.nonzero(existence)[0]:
    if i==0:
        vector = np.full(205, '    0')
        # Set the first entry to 10
        vector[0] = str(int(i))
        rows.insert(0,vector)
    if i>0:
        search = str(i-1)
        index = int(check_strings_in_csv(rows, search))+1# find which row to place the nan row
        vector = np.full(205, '    0')
        # Set the first entry to 0
        vector[0] = str(int(i))
        rows.insert(index,vector)
        
```

```python
# save the file including point coordinates of all frames
df_corr = pd.DataFrame(rows)
df_corr.to_csv(csvFile_Out+"_corrected.csv", index=False, na_rep='NaN',header=False) 
```

## 3.2. Update h5 file obtained from DLC project and close the h5 file

```python
#update datasets of the h5 file to include the imported annotations
new_image_count = 11900  # Total number of frames in behavioral recording
current_length = 5  #length of imgfiles number suffix. matching DLC extracted frames. usually 5

# Generate new values for each dataset
new_axis1_level2 = np.array([f'img{str(i).zfill(current_length)}.png'.encode('utf-8') for i in range(new_image_count)])
new_axis1_label0 = np.zeros(new_image_count, dtype=int)
new_axis1_label1 = np.zeros(new_image_count, dtype=int)
new_axis1_label2 = np.arange(new_image_count)

new_coordinates = rowsArr_floatxy[:new_image_count,:].transpose()
with h5py.File(H5DLC, 'a') as h5_file:
    # Delete existing datasets if needed
    for name, data in zip(['axis1_label0', 'axis1_label1', 'axis1_label2','axis1_level2'], 
                          [new_axis1_label0, new_axis1_label1, new_axis1_label2, new_axis1_level2]):
        
        if name in h5_file['keypoints'].keys():
            del h5_file[f'keypoints/{name}']
            print("deleted")
        
        # Create new resizable datasets
        h5_file.create_dataset(f'keypoints/{name}', data=data, maxshape=(None,))
    
# Open the H5 file and update `block0_values`
with h5py.File(H5DLC, 'a') as h5_file:
    block0_values = h5_file['keypoints/block0_values']
    
     # Delete existing dataset if needed
    if 'block0_values' in h5_file['keypoints'].keys():
        del h5_file['keypoints/block0_values']

    # Create a new dataset with resizable rows
    block0_values = h5_file.create_dataset(
        'keypoints/block0_values',
        data=new_coordinates,
        maxshape=(None, new_coordinates.shape[1])  # Allows resizing along rows
    )

#updatae the attributes of datasets that you updataes
for dataname in ['axis1_label0', 'axis1_label1', 'axis1_label2','axis1_level2']:
    for attr in hdold[f'keypoints/{dataname}'].attrs.keys():
        hd[f'keypoints/{dataname}'].attrs[attr] = hdold[f'keypoints/{dataname}'].attrs[attr]
```

```python
hd.close()
hdold.close()
```
